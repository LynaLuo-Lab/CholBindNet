```python
import glob
import os
import shutil

source_dir = "/home/alexhernandez/transmembranebindingAI/Data/SplitData/Cholesterol/cholesterol-graph-5A"
dest_dir = "/home/alexhernandez/CholBindNet/CLR_Ligand_Data/cholesterol-rdkit-vina-5A_exp1"
```

```python
spies = glob.glob(f"{source_dir}/Spies/*.npy")
positive_export_folder = glob.glob(f"{dest_dir}/PositiveWithoutSpies/*.npy")

for spy in spies:
    filename = os.path.basename(spy)
    spy_name = filename[:4]

    for positive in positive_export_folder:
        if spy_name in positive:
            destination_path = f"{dest_dir}/Spies/{os.path.basename(positive)}"
            os.makedirs(os.path.dirname(destination_path), exist_ok=True)
            shutil.move(positive, destination_path)
```

```python
import random

source_folder = f"{dest_dir}/UnlabeledWithSpies"
k_subsets_folder = f"{dest_dir}/k_subsets"

# Make sure the k_subsets folder exists
os.makedirs(k_subsets_folder, exist_ok=True)

# Get the list of all files in the source folder
all_files = os.listdir(source_folder)

# Create 50 directories (k_subsets)
k = 50
for i in range(k):
    subset_folder = os.path.join(k_subsets_folder, f'subset_{i+1}')
    os.makedirs(subset_folder, exist_ok=True)

# Distribute 385 files to each subset 
for i in range(k):
    # Randomly select 385 files for this subset 
    selected_files = random.sample(all_files, k=(385*1))
    
    # Create the subset folder
    subset_folder = os.path.join(k_subsets_folder, f'subset_{i+1}')
    
    # Copy the selected files into the corresponding subset folder
    for file in selected_files:
        src_path = os.path.join(source_folder, file)
        dest_path = os.path.join(subset_folder, file)
        
        # Copy the file
        shutil.copy(src_path, dest_path)
        print(f'Copied {file} to {subset_folder}')

print(f'Finished copying 385 files into each of the {k} subsets.')
```
